### *“This* program reads a text string and analyses word frequencies ….”

```python
import string

text_string = "The rain fell on the car. The rain fell on the ground."
print(f'Text : {text_string}')

# Lowercase and strip punctuation
clean_text = text_string.lower().translate(
    str.maketrans("", "", string.punctuation)
)

print(f"\nOriginal : {text_string}")
print(f"Cleaned  : {clean_text}\n")

# Turn the text into a list of words.
split_text = clean_text.split()
print (f'Text after split : {split_text}')


word_frequencies = {}

for word in split_text:
  if word in word_frequencies:
    word_frequencies[word] += 1
  else:
    word_frequencies[word] = 1

print("\nWord frequencies:", word_frequencies)
```
